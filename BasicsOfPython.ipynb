7.What is the difference between “==” and “is” operators in Python ?

Ans. Equality operator check the value of two object are same or not (in
     memory).
     While identity operator 'is' check if two variable are refer the same object in memory. It only compares memory locations.





```python
x = [1,2,3] # Both lists have different memory location although they have same value.
y = [1,2,3]
print(x is y) # False because both lists have different memory location
print(x == y) # True  because both contain same value.
```

13. What is the difference between for and while loops ?

Ans. for loop :Used when the number of iterations is known.
               Iterates over a sequence (list, range, etc.).
               Iterate a fixed number of element.
    while loop : Used when iterations depends on a condition.
                 Runs until a condition becomes false.
                 We can also use while loop for sequence and collections but for loop are simpler .

14.Describe a scenario where a while loop is more suitable than a for loop.

Ans.Scenario : Imagine we need to keep asking a user to valid password     until they enter the correct one. Here the number of attempts is unknown, so we use while loop rather than for loop.

```python
#7.Basic arithmatic operation on variables
# Get user input
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
if num2 != 0:
    division = num1 / num2
else:
    division = "undefined (cannot divide by zero)"

# Display results
print(f"Addition: {num1} + {num2} = {addition}")
print(f"Subtraction: {num1} - {num2} = {subtraction}")
print(f"Multiplication: {num1} * {num2} = {multiplication}")
print(f"Division: {num1} / {num2} = {division}")
```

```python
#9. Write a Python program to convert user input from string to integer, float, and boolean types.
user_input = input("Enter a string: ")

# Convert to integer
try:
    int_value = int(user_input)
    print("Integer value:", int_value)
except ValueError:
    print("Cannot convert to integer")

# Convert to float
try:
    float_value = float(user_input)
    print("Float value:", float_value)
except ValueError:
    print("Cannot convert to float")

# Convert to boolean
bool_value = bool(user_input)
print("Boolean value:", bool_value)
```

```python
#13. Sum of all even number between 1 to 50
even_sum = 0
for num in range(1, 51):
    if num % 2 == 0:
        even_sum += num

print("Sum of all even numbers between 1 and 50:", even_sum)
```

```python
#14.Get user input
user_string = input("Enter a string: ")

reversed_string = ""
index = len(user_string) - 1

# Reverse the string using a while loop
while index >= 0:
    reversed_string += user_string[index]
    index -= 1

print("Reversed string:", reversed_string)
```

```python
#15 Get user input
num = int(input("Enter a number: "))
factorial = 1

# Calculating the factorial using a while loo
while num > 0:
    factorial *= num
    num -= 1

print("Factorial:", factorial)
```
